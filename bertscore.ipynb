```python
import pandas as pd
from bert_score import score as bertscore

# Load predictions
df = pd.read_csv("predictions_wbc_urdu.csv")

# Remove rows with missing values
df = df.dropna(subset=["prediction", "reference"])

# Convert to strings and strip whitespace
predictions = df["prediction"].astype(str).str.strip().tolist()
references = df["reference"].astype(str).str.strip().tolist()

# Keep only valid prediction-reference pairs
filtered = [
    (p, r)
    for p, r in zip(predictions, references)
    if p != "" and r != ""
]

predictions = [p for p, r in filtered]
references = [r for p, r in filtered]

print("Number of samples:", len(predictions))
assert len(predictions) == len(references)

# Compute BERTScore
P, R, F1 = bertscore(
    predictions,
    references,
    lang="en",
    model_type="roberta-large",
    batch_size=8,
    verbose=True,
)

print("\n===== BERTScore =====")
print(f"Precision: {P.mean().item():.4f}")
print(f"Recall   : {R.mean().item():.4f}")
print(f"F1 Score : {F1.mean().item():.4f}")
```
